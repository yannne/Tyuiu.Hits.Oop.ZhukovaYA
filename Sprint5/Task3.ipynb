Создать базовый класс BankAccount в C#, который будет представлять
информацию об учетных записях в банке. На основе этого класса разработать 2-3
производных класса, демонстрирующих принципы наследования и полиморфизма.
В каждом из классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.

Требования к базовому классу BankAccount:

• Атрибуты: 
Номер счета (AccountNumber), Баланс (Balance), Тип счета (AccountType).

• Методы:

o GetInfo(): метод для получения информации о счете в виде строки.

o Deposit(decimal amount): метод для внесения денег на счет.

o Withdraw(decimal amount): метод для снятия денег со счета.

Требования к производным классам:

1. Сберегательный счет (SavingsAccount):
Должен содержать дополнительные атрибуты, такие как Процентная ставка (InterestRate). 
Метод Deposit() должен быть переопределен для добавления процентов к сумме вклада при внесении денег на счет.

2. Текущий счет (CheckingAccount): 
Должен содержать дополнительные атрибуты, такие как Лимит овердрафта (OverdraftLimit).
Метод Withdraw() должен быть переопределен для проверки и применения лимита овердрафта при снятии денег со счета.

3. Инвестиционный счет (InvestmentAccount) (если требуется третий класс):
Должен содержать дополнительные атрибуты, такие как Список активов (AssetsList). 
Метод GetInfo() должен быть переопределен для включения информации о списках активов в описании счета

```
using System;
using System.Collections.Generic;
using System.Linq;

// Делегаты
public delegate void AccountOperationDelegate(string message);
public delegate void TransactionHandler(decimal amount, string transactionType);

public interface ITransactionHistory
{
    void ShowTransactionHistory();
    void AddTransaction(string transaction);
}

public interface IAccountNotifications
{
    void SubscribeToNotifications();
    void SendNotification(string message);
}

// Интерфейс для наблюдателей
public interface IAccountObserver
{
    void OnAccountOperation(string accountNumber, string operation, decimal amount);
}

// Класс наблюдателя
public class AccountMonitor : IAccountObserver
{
    public string Name { get; set; }

    public AccountMonitor(string name)
    {
        Name = name;
    }

    public void OnAccountOperation(string accountNumber, string operation, decimal amount)
    {
        Console.WriteLine($"[Монитор {Name}] Счет {accountNumber}: операция '{operation}' на сумму {amount}");
    }
}

public class BankAccount : ITransactionHistory, IAccountNotifications
{
    private string _accountNumber;
    private decimal _balance = 0;
    private string _accountType;
    private DateTime _openingDate;
    private string _currency; 
    private bool _isActive;
    
    private string _ownerName;
    private string _email;
    private List<string> _transactionHistory;
    private bool _notificationsEnabled;

    // События
    public event AccountOperationDelegate AccountOperationPerformed;
    public event TransactionHandler TransactionCompleted;

    // Коллекция для хранения наблюдателей
    private List<IAccountObserver> _observers = new List<IAccountObserver>();

    public string AccountNumber{
        get {return _accountNumber;}
        set {_accountNumber = value;}
    }

    protected decimal Balance{
        get {return _balance;}
        set {_balance = value;}
    }

    protected string AccountType{
        get {return _accountType;}
        set {_accountType = value;}
    }

    protected DateTime OpeningDate{
        get {return _openingDate;}
        set {_openingDate = value;}
    }

    protected string Currency{
        get {return _currency;}
        set {_currency = value;}
    }

    protected bool IsActive{
        get {return _isActive;}
        set {_isActive = value;}
    }
    
    public string OwnerName {
        get { return _ownerName; }
        set { _ownerName = value; }
    }
    
    public string Email {
        get { return _email; }
        set { _email = value; }
    }
    
    protected List<string> TransactionHistory {
        get { return _transactionHistory; }
    }
    
    protected bool NotificationsEnabled {
        get { return _notificationsEnabled; }
        set { _notificationsEnabled = value; }
    }

    public BankAccount(string accountNumber, string accountType, string ownerName, string email){
        AccountNumber = accountNumber;
        AccountType = accountType;
        OpeningDate = DateTime.Now;
        Currency = "RUB";
        IsActive = true;
        OwnerName = ownerName;
        Email = email;
        _transactionHistory = new List<string>();
        _notificationsEnabled = false;
    }

    public void AddObserver(IAccountObserver observer)
    {
        _observers.Add(observer);
        Console.WriteLine($"Добавлен наблюдатель для счета {AccountNumber}");
    }

    public void RemoveObserver(IAccountObserver observer)
    {
        _observers.Remove(observer);
        Console.WriteLine($"Удален наблюдатель для счета {AccountNumber}");
    }

    private void NotifyObservers(string operation, decimal amount)
    {
        foreach (var observer in _observers)
        {
            observer.OnAccountOperation(AccountNumber, operation, amount);
        }
    }

    public void EnableNotifications()
    {
        _notificationsEnabled = true;
        Console.WriteLine($"Уведомления для счета {AccountNumber} включены");
    }
    
    public void DisableNotifications()
    {
        _notificationsEnabled = false;
        Console.WriteLine($"Уведомления для счета {AccountNumber} отключены");
    }
    
    public void ShowOwnerInfo()
    {
        Console.WriteLine($"Владелец: {OwnerName} | Email: {Email}");
    }
    
    protected void AddToTransactionHistory(string transaction)
    {
        string timestamp = DateTime.Now.ToString("dd.MM.yyyy HH:mm");
        _transactionHistory.Add($"[{timestamp}] {transaction}");
    }

    public virtual void GetInfo(){
        Console.WriteLine($"Номер счёта: {AccountNumber} | Баланс: {Balance} руб. | Тип счёта: {AccountType}");
        Console.WriteLine($"Дата открытия: {OpeningDate:dd.MM.yyyy} | Активен: {(IsActive ? "Да" : "Нет")}");
        ShowOwnerInfo();
    }

    public virtual void Deposit(decimal amount){
        if (!IsActive)
        {
            Console.WriteLine("Ошибка: счет закрыт");
            return;
        }
        Balance += amount;
        AddToTransactionHistory($"Пополнение: +{amount} {Currency}");
        
        AccountOperationPerformed?.Invoke($"Пополнение на {amount} {Currency}");
        TransactionCompleted?.Invoke(amount, "Deposit");
        NotifyObservers("Deposit", amount);
        
        if (_notificationsEnabled)
            ((IAccountNotifications)this).SendNotification($"Счет пополнен на {amount} {Currency}");
        Console.WriteLine($"Зачисление на счёт [{AccountNumber}] {amount} руб. Баланс: {Balance} руб.");
    }

    public virtual void Deposit(decimal amount, string message){
        if (!IsActive)
        {
            Console.WriteLine("Ошибка: счет закрыт");
            return;
        }
        Balance += amount;
        AddToTransactionHistory($"Пополнение: +{amount} {Currency} - {message}");
        
        AccountOperationPerformed?.Invoke($"Пополнение на {amount} {Currency}: {message}");
        TransactionCompleted?.Invoke(amount, "Deposit");
        NotifyObservers("Deposit", amount);
        
        if (_notificationsEnabled)
            ((IAccountNotifications)this).SendNotification($"Счет пополнен на {amount} {Currency}: {message}");
        Console.WriteLine($"Зачисление на счёт [{AccountNumber}] {amount} руб. Баланс: {Balance} руб.\nСообщение: \"{message}\"");
    }

    public virtual void Withdraw(decimal amount){
        if (!IsActive)
        {
            Console.WriteLine("Ошибка: счет закрыт");
            return;
        }
        if (Balance < amount) Console.WriteLine($"Недостаточность средств");
        else {
            Balance-= amount;
            AddToTransactionHistory($"Снятие: -{amount} {Currency}");
            
            AccountOperationPerformed?.Invoke($"Снятие {amount} {Currency}");
            TransactionCompleted?.Invoke(amount, "Withdraw");
            NotifyObservers("Withdraw", amount);
            
            if (_notificationsEnabled)
                ((IAccountNotifications)this).SendNotification($"Со счета снято {amount} {Currency}");
            Console.WriteLine($"Списание с счёта [{AccountNumber}] {amount} руб. Баланс: {Balance} руб.");
        }
    }

    public virtual void Transfer(BankAccount TransferAccount, decimal amout){
        if (!IsActive)
        {
            Console.WriteLine("Ошибка: счет закрыт");
            return;
        }
        if (amout <= 0) {
            Console.WriteLine($"Сумма перевода не может быть отрицательной или равна нулю"); return;}
        if (Balance < amout) {
            Console.WriteLine($"Недостаточно средств для перевода"); return;}
        Withdraw(amout);
        TransferAccount.Deposit(amout);
        AddToTransactionHistory($"Перевод на счет {TransferAccount.AccountNumber}: -{amout} {Currency}");

        AccountOperationPerformed?.Invoke($"Перевод {amout} {Currency} на счет {TransferAccount.AccountNumber}");
        NotifyObservers("Transfer", amout);
        
        Console.WriteLine($"Перевод средств на банкавский счет [{TransferAccount.AccountNumber}] завершен успешно");  
    }

    void ITransactionHistory.ShowTransactionHistory()
    {
        Console.WriteLine($"\n--- История операций по счету {AccountNumber} ---");
        if (_transactionHistory.Count == 0)
        {
            Console.WriteLine("Операций не было");
            return;
        }
        
        var sortedTransactions = _transactionHistory.OrderBy(t => t).ToList();
        foreach (var transaction in sortedTransactions)
        {
            Console.WriteLine(transaction);
        }
    }
    
    void ITransactionHistory.AddTransaction(string transaction)
    {
        AddToTransactionHistory(transaction);
    }

    void IAccountNotifications.SubscribeToNotifications()
    {
        EnableNotifications();
    }
    
    void IAccountNotifications.SendNotification(string message)
    {
        if (_notificationsEnabled)
        {
            Console.WriteLine($"УВЕДОМЛЕНИЕ для {OwnerName}: {message}");
        }
    }

    public virtual void CloseAccount()
    {
        if (Balance == 0)
        {
            IsActive = false;
            AddToTransactionHistory("Счет закрыт");
            
            // Событие закрытия счета
            AccountOperationPerformed?.Invoke("Счет закрыт");
            NotifyObservers("Close", 0);
            
            Console.WriteLine($"Счет [{AccountNumber}] закрыт");
        }
        else
        {
            Console.WriteLine("Невозможно закрыть счет: баланс не равен нулю");
        }
    }

    public void GetAccountAge()
    {
        TimeSpan age = DateTime.Now - OpeningDate;
        Console.WriteLine($"Счет открыт {age.Days} дней назад");
    }

    public virtual void CalculateFees()
    {
        Console.WriteLine("Комиссии не предусмотрены для данного типа счета");
    }
}

public class Bank
{
    private List<BankAccount> _accounts = new List<BankAccount>();
    private Dictionary<string, BankAccount> _accountDictionary = new Dictionary<string, BankAccount>();

    public void AddAccount(BankAccount account)
    {
        _accounts.Add(account);
        _accountDictionary[account.AccountNumber] = account;
        Console.WriteLine($"Счет {account.AccountNumber} добавлен в банк");
    }

    public BankAccount FindAccount(string accountNumber)
    {
        return _accountDictionary.ContainsKey(accountNumber) ? _accountDictionary[accountNumber] : null;
    }

    public void ProcessAllAccounts(Action<BankAccount> accountAction)
    {
        Console.WriteLine($"\n--- Обработка всех счетов ({_accounts.Count} шт.) ---");
        _accounts.ForEach(accountAction);
    }

    public IEnumerable<BankAccount> GetAccountsByOwner(string ownerName)
    {
        return _accounts.Where(acc => acc.OwnerName.Contains(ownerName));
    }

    public void ShowTotalBalance()
    {
        decimal total = _accounts.Sum(acc => 
        {
            var balanceProperty = acc.GetType().GetProperty("Balance", 
                System.Reflection.BindingFlags.NonPublic | System.Reflection.BindingFlags.Instance);
            return balanceProperty != null ? (decimal)balanceProperty.GetValue(acc) : 0;
        });
        
        Console.WriteLine($"Общий баланс всех счетов: {total} RUB");
    }

    public void ProcessLargeAccounts(decimal minBalance)
    {
        var largeAccounts = _accounts.Where(acc => 
        {
            var balanceProperty = acc.GetType().GetProperty("Balance", 
                System.Reflection.BindingFlags.NonPublic | System.Reflection.BindingFlags.Instance);
            return balanceProperty != null && (decimal)balanceProperty.GetValue(acc) >= minBalance;
        });

        Console.WriteLine($"\n--- Счета с балансом >= {minBalance} RUB ---");
        foreach (var account in largeAccounts)
        {
            account.GetInfo();
        }
    }
}

public class AccountService
{
    private readonly BankAccount _account;

    public AccountService(BankAccount account)
    {
        _account = account;
    }

    public void PerformAccountOperations()
    {
        Console.WriteLine($"\n--- Работа с счетом {_account.AccountNumber} ---");
        _account.GetInfo();
        _account.Deposit(1000);
        _account.GetAccountAge();
        
        ((ITransactionHistory)_account).ShowTransactionHistory();
    }
}

public class SavingsAccount : BankAccount
{
    private decimal _interestRate;
    private decimal _minBalance;
    private DateTime _lastInterestCalculation;

    private decimal InterestRate{
        get {return _interestRate;}
        set {_interestRate = value;}
    }

    private decimal MinBalance{
        get {return _minBalance;}
        set {_minBalance = value;}
    }

    public SavingsAccount(string accountNumber, decimal interestRate, string ownerName, string email) 
        : base(accountNumber, "Сберегательный счёт", ownerName, email)
    {
        InterestRate = interestRate;
        MinBalance = 100;
        _lastInterestCalculation = DateTime.Now;
    }
    
    public void CalculateMonthlyInterest()
    {
        if ((DateTime.Now - _lastInterestCalculation).TotalDays >= 30)
        {
            var balanceProperty = GetType().GetProperty("Balance", 
                System.Reflection.BindingFlags.NonPublic | System.Reflection.BindingFlags.Instance);
            decimal currentBalance = balanceProperty != null ? (decimal)balanceProperty.GetValue(this) : 0;
            
            decimal interest = currentBalance * InterestRate / 100 / 12;
            
            var currencyProperty = GetType().GetProperty("Currency", 
                System.Reflection.BindingFlags.NonPublic | System.Reflection.BindingFlags.Instance);
            string currency = currencyProperty != null ? (string)currencyProperty.GetValue(this) : "RUB";
            
            balanceProperty.SetValue(this, currentBalance + interest);
            _lastInterestCalculation = DateTime.Now;
            ((ITransactionHistory)this).AddTransaction($"Начислены проценты: +{interest} {currency}");
            
            var operationEvent = GetType().GetField("AccountOperationPerformed", 
                System.Reflection.BindingFlags.Instance | System.Reflection.BindingFlags.NonPublic)?.GetValue(this) as AccountOperationDelegate;
            operationEvent?.Invoke($"Начислены проценты: {interest} {currency}");
            
            Console.WriteLine($"Начислены ежемесячные проценты: {interest} {currency}");
        }
    }
    
    public override void Deposit(decimal amount)
    {
        var balanceProperty = GetType().GetProperty("Balance", 
            System.Reflection.BindingFlags.NonPublic | System.Reflection.BindingFlags.Instance);
        decimal currentBalance = balanceProperty != null ? (decimal)balanceProperty.GetValue(this) : 0;
        
        var currencyProperty = GetType().GetProperty("Currency", 
            System.Reflection.BindingFlags.NonPublic | System.Reflection.BindingFlags.Instance);
        string currency = currencyProperty != null ? (string)currencyProperty.GetValue(this) : "RUB";
        
        decimal interest = (amount * InterestRate) / 100;
        balanceProperty.SetValue(this, currentBalance + amount + interest);
        
        ((ITransactionHistory)this).AddTransaction($"Пополнение с процентами: +{amount + interest} {currency}");
        
        var operationEvent = GetType().GetField("AccountOperationPerformed", 
            System.Reflection.BindingFlags.Instance | System.Reflection.BindingFlags.NonPublic)?.GetValue(this) as AccountOperationDelegate;
        operationEvent?.Invoke($"Пополнение с процентами на {amount + interest} {currency}");
        
        var transactionEvent = GetType().GetField("TransactionCompleted", 
            System.Reflection.BindingFlags.Instance | System.Reflection.BindingFlags.NonPublic)?.GetValue(this) as TransactionHandler;
        transactionEvent?.Invoke(amount + interest, "Deposit with interest");
        
        Console.WriteLine($"Зачисление на счёт [{AccountNumber}] {amount} {currency} + проценты {interest} {currency}. Баланс: {currentBalance + amount + interest} {currency}");
    }

    public override void GetInfo(){
        base.GetInfo();
        Console.WriteLine($"Процентная ставка: {InterestRate} % | Мин. баланс: {MinBalance} {GetType().GetProperty("Currency", System.Reflection.BindingFlags.NonPublic | System.Reflection.BindingFlags.Instance)?.GetValue(this) ?? "RUB"}");
    }
}

        SavingsAccount savings = new SavingsAccount("5555", 5, "Кристина Ф.П.", "kristina@mail.ru");
        
        savings.AccountOperationPerformed += (message) => 
            Console.WriteLine($"[СОБЫТИЕ] {message}");
        
        savings.TransactionCompleted += (amount, type) => 
            Console.WriteLine($"[ТРАНЗАКЦИЯ] Тип: {type}, Сумма: {amount}");

        AccountMonitor monitor = new AccountMonitor("Главный");
        savings.AddObserver(monitor);

        Bank bank = new Bank();
        bank.AddAccount(savings);
        bank.AddAccount(new SavingsAccount("7777", 4, "Кристина Ф.П.", "kristina2@mail.ru"));
        Console.WriteLine("\n=== ОБРАБОТКА ВСЕХ СЧЕТОВ ===");
        bank.ProcessAllAccounts(acc => {
            Console.WriteLine($"--- Обработка счета: {acc.AccountNumber} ---");
            acc.GetInfo();
        });

        Console.WriteLine("\n=== СЧЕТА КРИСТИНЫ (LINQ) ===");
        var kristinaAccounts = bank.GetAccountsByOwner("Кристина");
        foreach (var account in kristinaAccounts)
        {
            account.GetInfo();
        }

        bank.ShowTotalBalance();

        bank.ProcessLargeAccounts(500);

        Console.WriteLine("\n=== ОПЕРАЦИИ СО СЧЕТОМ ===");
        AccountService savingsService = new AccountService(savings);
        savingsService.PerformAccountOperations();

        ((IAccountNotifications)savings).SubscribeToNotifications();
        savings.Deposit(300, "Зарплата");
        savings.Withdraw(200);
        
        ((ITransactionHistory)savings).ShowTransactionHistory();

        Console.WriteLine("\n=== НАЧИСЛЕНИЕ ПРОЦЕНТОВ ===");
        savings.CalculateMonthlyInterest();
        savings.CloseAccount();
```
